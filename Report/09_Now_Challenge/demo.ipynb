```python
import pandas as pd

# ví dụ hai bảng (DataFrame)
df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["A", "B", "C"]
})
df2 = pd.DataFrame({
    "id": [2, 3, 4],
    "score": [90, 80, 70]
})

# inner join (giống SQL INNER JOIN)
df_inner = pd.merge(df1, df2, on="id", how="inner")

# left join (giống SQL LEFT JOIN)
df_left = pd.merge(df1, df2, on="id", how="left")

# full outer join
df_full = pd.merge(df1, df2, on="id", how="outer")

# right join
df_right = pd.merge(df1, df2, on="id", how="right")
```
